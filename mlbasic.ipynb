```python
from sklearn.linear_model import LinearRegression

# X must be 2D: [[hours], [hours], ...] 
X = [[1], [2], [3], [4], [5],[6]]    
y = [16, 54, 65, 70, 81,67]         #3 cleaned data

model = LinearRegression() # training part/fit part
model.fit(X, y)                 # learn the pattern

prediction = model.predict([[7]])
print("Predicted score:", prediction[0])
```

```python
from sklearn.linear_model import LogisticRegression

X = [[1], [2], [3], [4], [5], [6]]
y = [1, 1, 0, 1, 1, 1]       # 0 = Fail, 1 = Pass

model = LogisticRegression()
model.fit(X, y) # training

new_student = [[4]]
prob_pass = model.predict_proba(new_student)[0]
predicted_class = model.predict(new_student)[0]

print("Probability of pass:", prob_pass)
print("Predicted class:", "Pass" if predicted_class == 1 else "Fail")
# if prob_pass >= 0.4:
#     print("Predicted class: Pass")
# else:
#     print("Predicted class: Fail")
```
